```python
import gradio as gr
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import re
import math
import time
import random
import torch
import os
import json
from datetime import datetime
from concurrent.futures import ThreadPoolExecutor
from functools import lru_cache
import requests
# Add these at the top of your code
import fitz  # PyMuPDF for PDF processing
import docx  # For Word document processing
from sklearn.feature_extraction.text import TfidfVectorizer
import io



# --------------------------
# UTILITY FUNCTIONS
# --------------------------

def extract_text_from_upload(uploaded_file):
    """Extract text from uploaded files - FIXED VERSION"""
    try:
        if uploaded_file is None:
            return ""
            
        print(f"Processing file: {uploaded_file.name if hasattr(uploaded_file, 'name') else 'unknown'}")
        
        # Read the file content
        if hasattr(uploaded_file, 'read'):
            # It's a file-like object
            file_content = uploaded_file.read()
            file_name = getattr(uploaded_file, 'name', '')
        else:
            # It might be a path string
            with open(uploaded_file, 'rb') as f:
                file_content = f.read()
            file_name = uploaded_file
        
        # Reset file pointer if possible
        if hasattr(uploaded_file, 'seek'):
            uploaded_file.seek(0)
            
        print(f"File size: {len(file_content)} bytes")
        
        # Process based on file extension
        if file_name.lower().endswith(('.txt', '.md', '.csv', '.html')):
            # Text files
            try:
                text = file_content.decode('utf-8')
                print(f"Extracted {len(text)} characters from text file")
                return text
            except UnicodeDecodeError:
                text = file_content.decode('latin-1')  # Fallback encoding
                print(f"Extracted {len(text)} characters from text file (latin-1)")
                return text
                
        elif file_name.lower().endswith('.pdf'):
            # PDF files
            try:
                # Create a BytesIO stream from the file content
                pdf_stream = io.BytesIO(file_content)
                doc = fitz.open(stream=pdf_stream, filetype="pdf")
                text = ""
                for page_num in range(len(doc)):
                    page = doc.load_page(page_num)
                    text += page.get_text()
                doc.close()
                print(f"Extracted {len(text)} characters from PDF")
                return text if text.strip() else "PDF appears to be empty or contains only images."
            except Exception as pdf_error:
                print(f"PDF extraction error: {pdf_error}")
                return f"Error reading PDF: {str(pdf_error)}. Using example data instead."
                
        elif file_name.lower().endswith(('.doc', '.docx')):
            # Word documents
            try:
                # Create a BytesIO stream from the file content
                doc_stream = io.BytesIO(file_content)
                doc = docx.Document(doc_stream)
                text = "\n".join([para.text for para in doc.paragraphs])
                print(f"Extracted {len(text)} characters from Word document")
                return text if text.strip() else "Document appears to be empty."
            except Exception as doc_error:
                print(f"Word document extraction error: {doc_error}")
                return f"Error reading Word document: {str(doc_error)}. Using example data instead."
        else:
            return f"Unsupported file format: {file_name}. Please upload a text, PDF, or Word document."
            
    except Exception as e:
        print(f"General file processing error: {e}")
        return f"Error processing file: {str(e)}. Using example data instead."

```

```python
def whitespace_tokenize_keep_punct(text):
    """Simple but effective tokenizer that preserves punctuation"""
    if not isinstance(text, str):
        return []
    # Replace certain punctuation with spaced versions to ensure they become separate tokens
    for c in '.,;:!?()[]{}<>':
        text = text.replace(c, f' {c} ')
    
    # Remove excessive whitespace and split
    return [t for t in text.split() if t]

def untokenize(tokens):
    """Convert tokens back to text, handling punctuation appropriately"""
    if not tokens:
        return ""
    
    punctuation = '.,;:!?()[]{}<>'
    result = tokens[0]
    
    for token in tokens[1:]:
        if token in punctuation:
            result += token
        else:
            result += ' ' + token
    
    return result
# --------------------------
# CREDIT SCORING FUNCTIONS
# --------------------------

import re

def get_score_from_finma_pipe(pipe, prompt):
    """Extract credit score robustly from LLM output, supporting multiple patterns and fallback."""
    try:
        if pipe is None:
            print("Warning: FinMA pipeline not available, using fallback scoring")
            return 650, "FinMA model not available. Using default score of 650."
        
        formatted_prompt = f"""
Analysis request: {prompt.strip()}

Instructions:
1. Analyze the credit risk in the above text
2. Return a credit score
3. Provide a brief explanation if needed

Credit score:"""
        
        output = pipe(formatted_prompt, max_new_tokens=50, do_sample=False, temperature=0.1)
        
        if isinstance(output, list) and len(output) > 0:
            output_text = output[0]['generated_text'].strip()
        else:
            return 650, "Error generating response. Using default score of 650."
        
        print(f"Model output: {output_text[:200]}...")
        
        # Normalize text to help matching (lowercase, strip)
        txt = output_text.lower()
        
        # Try various patterns
        
        # 1. Direct score patterns e.g. "Credit score: 700"
        patterns = [
            r"(\d{3})$",                                              # score at end of text
            r"credit score[^0-9]{0,10}(\d{3})",                       # Credit score: 700
            r"score[^0-9]{0,10}(\d{3})",                              # score: 700
            r"risk score[^0-9]{0,10}(\d{3})",                         # risk score: 700
            r"score of (\d{3})",                                      # score of 700
            r"(\d{3}) out of 850",                                    # 700 out of 850
            r"(\d{3}) points",                                        # 700 points
            
        ]
        
        for pattern in patterns:
            match = re.search(pattern, txt)
            if match:
                score_val = int(match.group(1))
                if 300 <= score_val <= 850:
                    print(f"Extracted score with pattern '{pattern}': {score_val}")
                    return score_val, output_text
        
        # 2. Percentage or out of 100 scale, convert to 300-850
        match = re.search(r"score[^0-9]{0,10}(\d{1,3})\s*\/\s*100", txt)
        if match:
            pct = int(match.group(1))
            score_val = int(300 + (pct * 5.5))
            score_val = min(850, max(300, score_val))
            print(f"Converted score from 100 scale: {pct} -> {score_val}")
            return score_val, output_text
        
        # 3. Score out of 10 scale
        match = re.search(r"score[^0-9]{0,10}(\d{1,2})\s*\/\s*10", txt)
        if match:
            val = int(match.group(1))
            score_val = int(300 + (val * 55))
            score_val = min(850, max(300, score_val))
            print(f"Converted score from 10 scale: {val} -> {score_val}")
            return score_val, output_text
        
        # 4. Words that indicate approval or rejection with a score estimate
        if "approve" in txt or "accepted" in txt or "granted" in txt:
            # Try to extract any number, else assign a good score
            found = re.findall(r"\b(\d{3})\b", txt)
            for num in found:
                val = int(num)
                if 300 <= val <= 850:
                    print(f"Approved with extracted score: {val}")
                    return val, output_text
            print("Approved but no score found; assigning 750")
            return 750, output_text
        
        if "reject" in txt or "denied" in txt or "decline" in txt:
            found = re.findall(r"\b(\d{3})\b", txt)
            for num in found:
                val = int(num)
                if 300 <= val <= 850:
                    print(f"Rejected with extracted score: {val}")
                    return val, output_text
            print("Rejected but no score found; assigning 400")
            return 400, output_text
        
        # 5. Fallback: extract any 3-digit number between 300 and 850 anywhere
        found_all = re.findall(r"\b(\d{3})\b", txt)
        for num in found_all:
            val = int(num)
            if 300 <= val <= 850:
                print(f"Fallback extracted score: {val}")
                return val, output_text
        
        print(f"Could not extract a valid score, returning default 650")
        return 650, output_text
    
    except Exception as e:
        print(f"Error in get_score_from_finma_pipe: {e}")
        return 650, f"Error in scoring: {str(e)}. Using default score of 650."

# Masked text handling
def is_heavily_masked(text, threshold=0.3):
    """Check if text contains too many mask tokens"""
    if not isinstance(text, str):
        return True
    
    # Also check for empty or very short text
    if len(text) < 10:
        return True
    
    mask_count = text.count('[MASK]')
    total_length = len(text.split())
    if total_length == 0:
        return True
        
    mask_ratio = mask_count / total_length
    
    # Also check for consecutive masks which often cause problems
    has_consecutive_masks = '[MASK] [MASK] [MASK]' in text
    
    return mask_ratio > threshold or has_consecutive_masks

def get_llm_score(prompt):
    """Get score from LLM with special handling for masked text"""
    try:
        # Skip heavily masked text - return neutral score
        if is_heavily_masked(prompt):
            return 650  # Return neutral score for heavily masked text
            
        score, _ = get_score_from_finma_pipe(finma_pipe, prompt)
        
        # Validate score is in reasonable range
        if score < 300 or score > 850:
            score = max(300, min(850, score))  # Clamp to valid range
            
        return score
    except Exception as e:
        # Only print error for non-masked text to reduce console spam
        if prompt.count('[MASK]') <= 5:
            print(f"Error in get_llm_score: {e}")
        return 650  # Default fallback score

# --------------------------
# MAIN PROCESSING FUNCTION - FIXED
# --------------------------
# Problem text detection
def is_problematic_text(text):
    """Enhanced detection for text patterns that will cause scoring problems"""
    if not isinstance(text, str):
        return True
        
    # Check for excessive masking
    mask_count = text.count('[MASK]')
    total_length = len(text.split())
    if total_length == 0 or mask_count / max(total_length, 1) > 0.2:
        return True
    
    # Check for consecutive masks which are problematic
    if '[MASK] [MASK] [MASK]' in text:
        return True
        
    # Check for HTML/CSS related content
    html_css_indicators = [
        '<', '>', 'style', 'href', 'body', 'font', 'link rel', 'content =',
        'text / css', 'table', 'wide', 'border'
    ]
    
    for indicator in html_css_indicators:
        if indicator in text:
            return True
            
    # Check for HTML-like structures or attributes
    if re.search(r'[<>{}]|=[^=]|\brel\b|\bhref\b|\bcontent\b|\bstyle\b|\bbody\b|\bfont\b', text):
        return True
        
    # Check for consecutive masked tokens or patterns with equals sign
    if re.search(r'\[MASK\]\s+\[MASK\]|\s+=\s+', text):
        return True
        
    return False
```

```python
# --------------------------
# ALIGNRAG IMPLEMENTATION
# --------------------------

class AlignRAG:
    """
    Aligned Retrieval-Augmented Generation for credit scoring.
    Retrieves relevant examples, policies, and standards to ensure
    consistent and explainable credit decisions.
    """
    
    def __init__(self, embedding_model=None, knowledge_base_path="knowledge_base"):
        self.embedding_model = embedding_model
        self.knowledge_base_path = knowledge_base_path
        self.knowledge_base = []
        self.embeddings = None
        self.vectorizer = None
        
        # Load knowledge base (credit policies, standards, precedents)
        self.load_knowledge_base()
        
       
    
    def load_knowledge_base(self):
        """Load credit knowledge base from files or create example data"""
        try:
            from sklearn.feature_extraction.text import TfidfVectorizer
            self.vectorizer = TfidfVectorizer(
                max_features=5000, 
                stop_words='english',
                ngram_range=(1, 2)
            )
            
            if os.path.exists(self.knowledge_base_path):
                # Load from disk
                files = os.listdir(self.knowledge_base_path)
                for file in files:
                    if file.endswith('.txt'):
                        with open(os.path.join(self.knowledge_base_path, file), 'r') as f:
                            content = f.read()
                            self.knowledge_base.append({
                                'id': file,
                                'content': content,
                                'type': self._determine_doc_type(file)
                            })
            
            # If no knowledge base exists or it's empty, create example data
            if not self.knowledge_base:
                print("Creating example knowledge base...")
                self._create_example_knowledge_base()
                
            # Create embeddings
            self._create_embeddings()
            
        except Exception as e:
            print(f"Error loading knowledge base: {e}")
            self._create_example_knowledge_base()
            self._create_embeddings()
    
    def _determine_doc_type(self, filename):
        """Determine document type from filename"""
        if 'policy' in filename.lower():
            return 'policy'
        elif 'standard' in filename.lower():
            return 'standard'
        elif 'precedent' in filename.lower():
            return 'precedent'
        elif 'regulation' in filename.lower():
            return 'regulation'
        else:
            return 'general'
    
    def _create_example_knowledge_base(self):
        """Create example knowledge base documents"""
        example_docs = [
            {
                'id': 'credit_policy_01.txt',
                'content': """
Credit Policy: Consumer Loan Assessment
- Income Verification: All income must be verified through paystubs or tax returns
- Debt-to-Income Ratio: Should not exceed 43% for approval
- Credit Score Requirements: 
  * 750+: Excellent - Lowest interest rates
  * 700-749: Good - Standard rates
  * 650-699: Fair - Higher rates, additional verification
  * Below 650: High Risk - May require collateral or cosigner
- Employment History: Minimum 2 years stable employment
- Prior Bankruptcies: 7-year waiting period required
                """,
                'type': 'policy'
            },
            {
                'id': 'credit_standards_income.txt',
                'content': """
Income Stability Standards:
- W2 Employment: Most stable, full weight given
- 1099 Contractor: 2+ years history required, 75% weight
- Self-Employment: 2+ years tax returns, 70% weight of stated income
- Investment Income: 3-year average, 50% weight
- Alimony/Child Support: Court documentation required, duration must exceed loan term
                """,
                'type': 'standard'
            },
            {
                'id': 'credit_precedent_medical_debt.txt',
                'content': """
Medical Debt Precedent:
Medical debt is treated with lower weight than consumer debt. Borrowers with primarily 
medical collections but otherwise good payment history may be approved with the 
following adjustments:
- Medical collections under $1,000: Minimal impact
- Medical collections $1,000-$5,000: 15-point score adjustment
- Medical collections over $5,000: 30-point score adjustment
                """,
                'type': 'precedent'
            },
            {
                'id': 'credit_regulation_fcra.txt',
                'content': """
Fair Credit Reporting Act Requirements:
- Adverse action notices must be provided when credit is denied
- Consumers must be informed of their right to access credit report
- Score factors affecting the decision must be disclosed
- Special circumstances (medical collections, identity theft) must be considered
- Consumers have right to dispute inaccurate information
                """,
                'type': 'regulation'
            },
            {
                'id': 'credit_example_excellent.txt',
                'content': """
Example: Excellent Credit Approval
Applicant profile:
- Credit score: 780
- Income: $85,000
- DTI: 28%
- Employment: 5 years at current employer
- No late payments in 24 months
- Savings: $30,000
- Existing debt: $15,000 mortgage, $5,000 auto loan
Decision: Approved at lowest interest rate tier
                """,
                'type': 'precedent'
            },
            {
                'id': 'credit_example_rejected.txt',
                'content': """
Example: Rejected Application
Applicant profile:
- Credit score: 580
- Income: $45,000
- DTI: 52%
- Employment: 8 months at current job, frequent job changes
- Multiple late payments in past 12 months
- Savings: $2,000
- Existing debt: $12,000 credit cards, $18,000 auto loan
Decision: Rejected due to high DTI and unstable employment
                """,
                'type': 'precedent'
            }
        ]
        
        self.knowledge_base = example_docs
    
    def _create_embeddings(self):
        """Create embeddings for knowledge base documents"""
        if not self.knowledge_base or not self.vectorizer:
            return
            
        documents = [doc['content'] for doc in self.knowledge_base]
        
        # If using a neural embedding model
        if self.embedding_model:
            try:
                # This would be replaced with actual embedding code
                self.embeddings = np.random.randn(len(documents), 768)
            except Exception:
                # Fallback to TF-IDF if embedding fails
                self.embeddings = self.vectorizer.fit_transform(documents)
        else:
            # Use TF-IDF embeddings
            self.embeddings = self.vectorizer.fit_transform(documents)
    
    def retrieve(self, query, top_k=3, filters=None):
        """
        Retrieve most relevant documents for a query with optional filtering
        """
        if not self.knowledge_base or self.embeddings is None or not self.vectorizer:
            return []
            
        # Create query embedding
        if self.embedding_model:
            try:
                # This would be replaced with actual embedding code
                query_embedding = np.random.randn(1, 768)
            except Exception:
                # Fallback to TF-IDF
                query_embedding = self.vectorizer.transform([query])
        else:
            query_embedding = self.vectorizer.transform([query])
            
        # Calculate similarities
        if isinstance(self.embeddings, np.ndarray):
            similarities = np.inner(query_embedding, self.embeddings)[0]
        else:
            similarities = np.array(query_embedding.dot(self.embeddings.T).toarray()[0])
            
        # Apply filters if specified
        filtered_indices = list(range(len(self.knowledge_base)))
        if filters:
            filtered_indices = []
            for i, doc in enumerate(self.knowledge_base):
                match = True
                for key, value in filters.items():
                    if key in doc and doc[key] != value:
                        match = False
                        break
                if match:
                    filtered_indices.append(i)
                    
        # Sort by similarity (only considering filtered documents)
        ranked_indices = sorted(
            [(i, similarities[i]) for i in filtered_indices],
            key=lambda x: x[1],
            reverse=True
        )
        
        # Get top-k documents
        results = []
        for idx, score in ranked_indices[:top_k]:
            doc = self.knowledge_base[idx].copy()
            doc['relevance_score'] = float(score)
            results.append(doc)
            
        return results
    
    def extract_credit_factors(self, text):
        """
        Extract key credit factors from application text
        """
        factors = {
            'income': None,
            'credit_score': None,
            'dti': None,
            'employment_years': None,
            'late_payments': None,
            'savings': None,
            'existing_debt': None
        }
        
        # Income extraction
        income_match = re.search(r'income:?\s*\$?(\d+[,\d]*)', text, re.IGNORECASE)
        if income_match:
            try:
                factors['income'] = float(income_match.group(1).replace(',', ''))
            except ValueError:
                pass
                
        # Credit score extraction
        score_match = re.search(r'(?:credit\s+score|fico):?\s*(\d{3})', text, re.IGNORECASE)
        if score_match:
            try:
                factors['credit_score'] = int(score_match.group(1))
            except ValueError:
                pass
                
        # DTI extraction
        dti_match = re.search(r'(?:dti|debt[\s-]*to[\s-]*income):?\s*(\d+(?:\.\d+)?)\%?', text, re.IGNORECASE)
        if dti_match:
            try:
                factors['dti'] = float(dti_match.group(1))
            except ValueError:
                pass
                
        # Employment years
        emp_match = re.search(r'(?:employment|job|position).{1,30}?(\d+)\s*years?', text, re.IGNORECASE)
        if emp_match:
            try:
                factors['employment_years'] = float(emp_match.group(1))
            except ValueError:
                pass
                
        # Late payments
        late_match = re.search(r'(\d+)\s*late\s*payments?', text, re.IGNORECASE)
        if late_match:
            try:
                factors['late_payments'] = int(late_match.group(1))
            except ValueError:
                pass
        else:
            # Check for "no late payments"
            no_late = re.search(r'no\s*late\s*payments?', text, re.IGNORECASE)
            if no_late:
                factors['late_payments'] = 0
                
        # Savings
        savings_match = re.search(r'savings:?\s*\$?(\d+[,\d]*)', text, re.IGNORECASE)
        if savings_match:
            try:
                factors['savings'] = float(savings_match.group(1).replace(',', ''))
            except ValueError:
                pass
                
        # Existing debt
        debt_match = re.search(r'(?:debt|loans?):?\s*\$?(\d+[,\d]*)', text, re.IGNORECASE)
        if debt_match:
            try:
                factors['existing_debt'] = float(debt_match.group(1).replace(',', ''))
            except ValueError:
                pass
                
        return factors
    
    def get_policy_alignment(self, factors):
        """
        Check how well application aligns with credit policies
        """
        alignment_score = 0.0
        explanations = []
        
        # Credit score alignment
        if factors['credit_score']:
            score = factors['credit_score']
            if score >= 750:
                alignment_score += 30
                explanations.append("Credit score is excellent (750+)")
            elif score >= 700:
                alignment_score += 25
                explanations.append("Credit score is good (700-749)")
            elif score >= 650:
                alignment_score += 15
                explanations.append("Credit score is fair (650-699)")
            else:
                explanations.append("Credit score is below standard (<650)")
                
        # DTI alignment
        if factors['dti']:
            dti = factors['dti']
            if dti <= 36:
                alignment_score += 25
                explanations.append("DTI ratio is excellent (≤36%)")
            elif dti <= 43:
                alignment_score += 15
                explanations.append("DTI ratio is acceptable (≤43%)")
            else:
                explanations.append("DTI ratio exceeds guidelines (>43%)")
                
        # Employment stability
        if factors['employment_years']:
            years = factors['employment_years']
            if years >= 2:
                alignment_score += 20
                explanations.append(f"Employment history meets guidelines ({years} years)")
            else:
                explanations.append(f"Employment history below 2-year guideline ({years} years)")
                
        # Payment history
        if factors['late_payments'] is not None:
            if factors['late_payments'] == 0:
                alignment_score += 15
                explanations.append("No late payments reported")
            elif factors['late_payments'] <= 1:
                alignment_score += 5
                explanations.append("Minimal late payments (1)")
            else:
                explanations.append(f"Multiple late payments ({factors['late_payments']})")
                
        # Normalize score to 0-100
        final_score = min(100, alignment_score)
        
        return final_score, explanations
    
    def generate_aligned_explanation(self, application_text, score):
        """
        Generate aligned explanation using retrieved knowledge
        """
        # Extract credit factors
        factors = self.extract_credit_factors(application_text)
        
        # Calculate policy alignment
        alignment_score, factor_explanations = self.get_policy_alignment(factors)
        
        # Retrieve relevant policies and precedents
        policies = self.retrieve(application_text, top_k=2, filters={'type': 'policy'})
        precedents = self.retrieve(application_text, top_k=1, filters={'type': 'precedent'})
        
        # Choose most relevant precedent based on score
        if score >= 700:
            target_examples = self.retrieve("excellent credit approval", top_k=1)
        elif score >= 600:
            target_examples = self.retrieve("good credit approval", top_k=1)
        else:
            target_examples = self.retrieve("rejected application", top_k=1)
        
        # Build explanation components
        explanation = {
            'credit_score': int(score),
            'policy_alignment': alignment_score,
            'timestamp': self.timestamp,
            'user': self.user,
            'factor_explanations': factor_explanations,
            'supporting_policies': [p['id'] for p in policies],
            'similar_precedents': [p['id'] for p in precedents + target_examples],
            'recommendation': "approve" if score >= 650 else "reject",
            'explanation_text': self._format_explanation(score, factors, alignment_score, 
                                                        factor_explanations, policies, precedents),
            'evidence': {
                'factors': factors,
                'policies': policies,
                'precedents': precedents
            }
        }
        
        return explanation
    
    def _format_explanation(self, score, factors, alignment_score, 
                           factor_explanations, policies, precedents):
        """Format the explanation text using retrieved evidence"""
        risk_level = "low" if score >= 750 else "medium" if score >= 650 else "high" if score >= 550 else "very high"
        
        explanation = f"""
Credit Assessment Result: {int(score)} (Policy Alignment: {alignment_score:.1f}%)
Risk Level: {risk_level.upper()} RISK

Key Factors:
{chr(10).join(['- ' + exp for exp in factor_explanations])}

"""
        # Add policy references if available
        if policies:
            explanation += "\nRelevant Policy Considerations:\n"
            for policy in policies:
                # Extract a relevant snippet
                content = policy['content']
                # Find most relevant section based on factors
                sections = content.split('\n- ')
                selected_section = sections[0]
                for section in sections:
                    if factors['credit_score'] and 'score' in section.lower():
                        selected_section = section
                        break
                    elif factors['dti'] and ('dti' in section.lower() or 'debt-to-income' in section.lower()):
                        selected_section = section
                        break
                
                explanation += f"- {selected_section.strip()}\n"
        
        # Add recommendation
        explanation += f"\nRecommendation: {'APPROVE' if score >= 650 else 'REJECT'}\n"
        
        # Add timestamp and audit info
        explanation += f"\nAssessment completed at {self.timestamp} by {self.user}"
        
        return explanation
```

```python
# Wrapper for HF model with batch processing support
from typing import Optional, Callable, List, Dict, Any
import numpy as np
import torch
from concurrent.futures import ThreadPoolExecutor


# Wrapper for HF model with batch processing support
class FastModelWrapper:
    """Optimized model wrapper with parallel processing and caching"""
    
    def __init__(self, model=None, tokenizer=None, score_fn=None, batch_size=8):
        self.model = model
        self.tokenizer = tokenizer
        self.score_fn = score_fn or get_llm_score
        self.batch_size = batch_size
        self.score_cache = {}  # In-memory cache
        self.max_workers = 4  # Number of parallel workers
    
    @lru_cache(maxsize=1024)  # Cache up to 1024 individual text scores
    def _score_single(self, text):
        """Score a single text with caching"""
        try:
            return float(self.score_fn(text))
        except Exception:
            return 650.0
    
    def _batch_score(self, texts):
        """Score a batch of texts efficiently"""
        if self.model is not None and self.tokenizer is not None:
            try:
                # Efficient HF model inference
                inputs = self.tokenizer(texts, return_tensors="pt", padding=True, truncation=True, max_length=512)
                inputs = {k: v.to(self.model.device) for k, v in inputs.items()}
                
                with torch.no_grad():
                    outputs = self.model(**inputs)
                
                if hasattr(outputs, "logits"):
                    # Extract and process logits efficiently
                    logits = outputs.logits.detach().cpu().numpy()
                    scores = np.max(logits, axis=-1).mean(axis=-1)
                    return scores.astype(float)
            except Exception:
                pass  # Fall back to individual scoring
        
        # Use ThreadPoolExecutor for parallel scoring
        with ThreadPoolExecutor(max_workers=self.max_workers) as executor:
            scores = list(executor.map(self._score_single, texts))
        return np.array(scores, dtype=float)
    
    def __call__(self, texts):
        """Process texts in optimized batches with parallel execution"""
        if not isinstance(texts, list):
            if isinstance(texts, str):
                texts = [texts]
            else:
                return np.array([650.0])
        
        if not texts:
            return np.array([], dtype=float)
            
        # Check cache first for all texts
        cached_scores = []
        uncached_texts = []
        uncached_indices = []
        
        for i, text in enumerate(texts):
            if text in self.score_cache:
                cached_scores.append((i, self.score_cache[text]))
            else:
                uncached_texts.append(text)
                uncached_indices.append(i)
        
        results = np.zeros(len(texts), dtype=float)
        
        # Fill in cached results
        for i, score in cached_scores:
            results[i] = score
            
        # Process uncached texts in optimized batches
        if uncached_texts:
            # Determine optimal batch size based on text length
            avg_length = sum(len(t) for t in uncached_texts) / len(uncached_texts)
            adaptive_batch_size = max(1, min(self.batch_size, int(4000 / avg_length)))
            
            for i in range(0, len(uncached_texts), adaptive_batch_size):
                batch = uncached_texts[i:i+adaptive_batch_size]
                batch_scores = self._batch_score(batch)
                
                # Update cache and results
                for j, score in enumerate(batch_scores):
                    text_idx = i + j
                    if text_idx < len(uncached_texts):
                        text = uncached_texts[text_idx]
                        self.score_cache[text] = float(score)
                        if text_idx < len(uncached_indices):
                            results[uncached_indices[text_idx]] = float(score)
        
        return results

def fast_tokenshap_analyze(model_wrapper, prompt, n_samples=10, mask_token="[MASK]"):
    """
    Ultra-fast TokenSHAP implementation using:
    1. Parallel processing
    2. Adaptive sampling
    3. Pruning with early stopping
    4. Incremental computation
    """
    # Clean and efficiently tokenize input
    # Use regex substitution for HTML cleanup in one pass
    clean_prompt = re.sub(r'<[^>]*>|[{}\[\]]', ' ', prompt)
    tokens = whitespace_tokenize_keep_punct(clean_prompt)
    
    if not tokens:
        return pd.DataFrame(columns=["token", "attribution", "abs_attrib"]), go.Figure()
    
    # Limit tokens but keep important ones
    max_tokens = 30
    if len(tokens) > max_tokens:
        # Keep first, last, and important-looking tokens
        important_pattern = re.compile(r'credit|loan|income|debt|score|payment|history|asset', re.IGNORECASE)
        importance_scores = []
        for i, token in enumerate(tokens):
            if i >= len(tokens):
                break
            
            importance = 0
            if important_pattern.search(token):
                importance = 3
            elif token and token[0].isupper():
                importance = 2
            elif token and token.isalnum():
                importance = 1
                
            importance_scores.append((i, importance))
        
        # Sort by importance score, then position
        importance_scores.sort(key=lambda x: (-x[1], abs(x[0] - len(tokens)/2)))
        
        # Get indices of tokens to keep
        keep_indices = sorted([i for i, _ in importance_scores[:max_tokens-6]])
        # Always include some context from beginning and end
        end_index = min(len(tokens)-1, max(0, len(tokens)-3))
        keep_indices = sorted(list(set(keep_indices + list(range(3)) + list(range(end_index, len(tokens))))))[:max_tokens]
        
        # Keep only selected tokens
        tokens = [tokens[i] for i in keep_indices if i < len(tokens)]
    
    n = len(tokens)
    print(f"Fast TokenSHAP on {n} tokens with up to {n_samples} samples")
    
    try:
        # Create optimized FastModelWrapper with larger batch size
        if not isinstance(model_wrapper, FastModelWrapper):
            fast_wrapper = FastModelWrapper(
                model=getattr(model_wrapper, 'model', None),
                tokenizer=getattr(model_wrapper, 'tokenizer', None),
                score_fn=getattr(model_wrapper, 'score_fn', None),
                batch_size=16
            )
        else:
            fast_wrapper = model_wrapper
        
        # Get baseline score
        baseline_text = " ".join(tokens)
        baseline_score = float(fast_wrapper([baseline_text])[0])
        
        # STEP 1: Quick initial evaluation of each token
        print("Quick initial token evaluation...")
        leave_one_out_texts = []
        for i in range(n):
            masked_tokens = tokens.copy()
            masked_tokens[i] = mask_token
            leave_one_out_texts.append(" ".join(masked_tokens))
        
        # Score all leave-one-out texts in one batch
        leave_one_out_scores = fast_wrapper(leave_one_out_texts)
        
        # Calculate initial attributions
        attributions = np.zeros(n)
        confidences = np.ones(n)
        
        for i, score in enumerate(leave_one_out_scores):
            attributions[i] = baseline_score - score
        
        # STEP 2: Identify important tokens for deeper analysis
        importance_threshold = np.percentile(np.abs(attributions), 70)
        important_indices = np.where(np.abs(attributions) > importance_threshold)[0]
        
        # STEP 3: Adaptive sampling for important tokens
        if len(important_indices) > 0 and n_samples > 1:
            print(f"Detailed analysis of {len(important_indices)} important tokens...")
            
            # Prepare all sampling texts in advance
            sample_texts = []
            sample_meta = []  # (token_idx, sample_idx, with_token)
            
            # For each important token, generate multiple random context samples
            samples_per_token = min(n_samples - 1, 4)
            for idx in important_indices:
                for s in range(samples_per_token):
                    # Create random context (keep 30-70% of tokens)
                    keep_ratio = 0.3 + (0.4 * np.random.random())
                    mask = np.random.random(n) < keep_ratio
                    
                    # Version without the token
                    mask[idx] = False
                    without_token = [tokens[j] if mask[j] else mask_token for j in range(n)]
                    without_text = " ".join(without_token)
                    sample_texts.append(without_text)
                    sample_meta.append((idx, s, 0))
                    
                    # Version with the token
                    mask[idx] = True
                    with_token = [tokens[j] if mask[j] else mask_token for j in range(n)]
                    with_text = " ".join(with_token)
                    sample_texts.append(with_text)
                    sample_meta.append((idx, s, 1))
            
            # Score all samples in optimized batches
            sample_scores = fast_wrapper(sample_texts)
            
            # Update attributions based on sample results
            for i, (token_idx, sample_idx, with_token) in enumerate(sample_meta):
                if i < len(sample_scores):
                    score = sample_scores[i]
                    
                    # Find the matching pair to calculate marginal contribution
                    pair_idx = -1
                    for j, (t_idx, s_idx, w_token) in enumerate(sample_meta):
                        if t_idx == token_idx and s_idx == sample_idx and w_token != with_token:
                            pair_idx = j
                            break
                    
                    if pair_idx >= 0 and pair_idx < len(sample_scores):
                        pair_score = sample_scores[pair_idx]
                        
                        # Calculate marginal contribution (with - without)
                        contribution = 0
                        if with_token == 1:
                            contribution = score - pair_score
                        else:
                            contribution = pair_score - score
                        
                        # Update running average
                        attributions[token_idx] = ((attributions[token_idx] * confidences[token_idx]) + 
                                                 contribution) / (confidences[token_idx] + 1)
                        confidences[token_idx] += 1
        
        # STEP 4: Normalize and format results
        max_abs = np.max(np.abs(attributions))
        if max_abs > 0:
            attributions = attributions / max_abs * 100
        
        # Create DataFrame and visualization
        df = pd.DataFrame({
            "token": tokens,
            "attribution": attributions,
            "abs_attrib": np.abs(attributions),
            "confidence": confidences
        })
        df = df.sort_values("abs_attrib", ascending=False).reset_index(drop=True)
        
        # Create visualization (show only top contributors for clarity)
        top_n = min(20, n)
        top_df = df.head(top_n)
        
        fig = go.Figure()
        colors = ['#22c55e' if v > 0 else '#ef4444' for v in top_df["attribution"]]
        fig.add_trace(go.Bar(
            x=top_df["abs_attrib"], 
            y=top_df["token"], 
            orientation='h',
            marker_color=colors, 
            text=[f"{a:+.1f}" for a in top_df["attribution"]],
            textposition='auto',
            hovertemplate="<b>%{y}</b><br>Attribution: %{text}<extra></extra>"
        ))
        
        fig.update_layout(
            title=f"Fast TokenSHAP Analysis ({n} tokens)", 
            xaxis_title="Relative Impact (%)",
            height=min(500, max(300, top_n * 25 + 100)),
            margin=dict(l=10, r=10, t=30, b=10)
        )
        
        return df, fig
        
    except Exception as e:
        print(f"Error in fast TokenSHAP analysis: {str(e)}")
        empty_df = pd.DataFrame({"token": ["Error"], "attribution": [0], "abs_attrib": [0]})
        fig = go.Figure()
        fig.update_layout(title="Analysis failed", height=300)
        return empty_df, fig



# --------------------------
# CREDIT SCORING DISPLAY
# --------------------------

def get_risk_level_and_color_from_score(score_val):
    if score_val >= 750:
        return "LOW RISK", "#22c55e", "Excellent"
    elif score_val >= 650:
        return "MEDIUM RISK", "#f59e0b", "Good"
    elif score_val >= 550:
        return "HIGH RISK", "#ef4444", "Fair"
    else:
        return "VERY HIGH RISK", "#dc2626", "Poor"

def format_score_display_html(score_val, decision):
    risk_level, color, rating = get_risk_level_and_color_from_score(score_val)
    recommendation = "PENDING"
    if isinstance(decision, str) and "approve" in decision.lower():
        recommendation = "APPROVED ✅"; rec_color = "#22c55e"
    elif isinstance(decision, str) and "reject" in decision.lower():
        recommendation = "REJECTED ❌"; rec_color = "#ef4444"
    else:
        rec_color = "#6b7280"
    
    score_int = int(round(score_val))
    dash = 2 * math.pi * 90
    offset = dash * (1 - float(score_int) / 1000.0)
    
    html = f"""
<div style="background: linear-gradient(135deg, #f8fafc, #f1f5f9); border-radius: 12px; padding: 18px; border: 1px solid #e2e8f0;">
  <div style="text-align:center;">
    <div style="font-size:20px; font-weight:700; color:#0f172a;">Credit Assessment</div>
    <div style="display:inline-block; margin-top:12px;">
      <svg width="160" height="160" style="transform: rotate(-90deg);">
        <circle cx="80" cy="80" r="70" fill="none" stroke="#e2e8f0" stroke-width="12"/>
        <circle cx="80" cy="80" r="70" fill="none" stroke="{color}" stroke-width="12"
                stroke-dasharray="{dash}" stroke-dashoffset="{offset}" stroke-linecap="round"/>
      </svg>
    </div>
    <div style="margin-top:8px; font-size:28px; font-weight:700; color:{color};">{score_int}</div>
    <div style="font-size:12px; color:#64748b;">out of 850 — {rating}</div>
    <div style="margin-top:10px; display:inline-block; background:{rec_color}; color:#fff; padding:8px 14px; border-radius:10px; font-weight:600;">{recommendation}</div>
  </div>
</div>
"""
    return html
```
